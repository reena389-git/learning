```python
# create secret for API using databricks workspace client
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()

# create a scope (ignore error if it already exists)
try:
    w.secrets.create_scope(scope="finnhub-api")
except Exception as e:
    print("scope may already exist:", e)

# store your key
w.secrets.put_secret(scope="finnhub", key="api_key", string_value="d9q7gr1r01qo4q2scer0d9q7gr1r01qo4q2scerg")
print("secret stored")

# verify it's there
print("keys in scope:", [s.key for s in w.secrets.list_secrets(scope="finnhub-api")])
```
